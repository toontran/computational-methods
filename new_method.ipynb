```python
import numpy as np
from scipy.linalg import eig, svd
from scipy.sparse.linalg import eigs, svds
from scipy.sparse.linalg import LinearOperator

np.random.seed(10)

# Parameters
N = 12  # Full matrix size
w = 4  # Window size
keep = w #// 2  # Number of components to keep (1)

# Generate full 4x4 SPD matrix for testing/comparison
Q = np.random.randn(N,N)
Q, _ = np.linalg.qr(Q)
S = np.linspace(1,N,N)**2
Q = Q[:, :4]
S = S[:4]
full_matrix = Q @ np.diag(S) @ Q.T
full_matrix = (full_matrix + full_matrix.T) / 2
print("Full matrix:")
print(full_matrix)

# First window (2x4)
first_window = full_matrix[:w, :]  

# Step 1: First window eigendecomposition (2x2)
S11, V11 = eig(first_window[:w, :w])
idx = np.argsort(S11)[::-1]
S11, V11 = S11[idx], V11[:, idx]

print("\nStep 1 results:")
print("S11:", S11)
print("V11:", V11)

# Step 2: Compute SVD of augmented matrix
S11_sqrt = np.sqrt(np.abs(S11))
S11_invsqrt = np.zeros_like(S11)
mask = S11 > 1e-10
S11_invsqrt[mask] = 1 / np.sqrt(S11[mask])

upper = V11 @ np.diag(S11_sqrt) @ V11.T
lower = first_window[:, w:].T @ V11 @ np.diag(S11_invsqrt) @ V11.T  # Empty for first window
aug_matrix = np.vstack([upper, lower])

U, S, Vt = svd(aug_matrix, full_matrices=False)
V = U[:, :keep]  # This V is now of shape (w, keep)
S = S[:keep]**2

print("\nStep 2 results:")
print("S:", S)
print("V shape:", V.shape)
# raise

# Second window (2x4)
second_window = full_matrix[w:2*w, :]

# Step 3: Expand A11 and use eigs
def matvec(x):
    x = x.reshape(-1)
    result = np.zeros(2*w, dtype=np.float64)
    
    # Upper left block: V S V^T
    result[:w] = (V[:w, :] @ (np.diag(S) @ (V[:w, :].T @ x[:w]))).real
    
    # Lower left block
    result[w:] = second_window[:, w:].T @ x[w:]
    
    return result

def make_operator(new_window, n, w, V, S):
    def matvec(x):
        x = x.reshape(-1)
        result = np.zeros(n*w, dtype=np.float64)

        # First (n-1)w rows: V S V^T part
        V_old = V[:(n-1)*w, :]  # All but last block
        x_old = x[:(n-1)*w]     # First (n-1)w elements

        # Compute V S V^T x for all old blocks at once
        result[:(n-1)*w] = V_old @ (S * (V_old.T @ x_old))

        # Last block: new window direct computation
        result[(n-1)*w:n*w] = new_window @ x[(n-1)*w:n*w]

        # Cross terms between new window and old blocks
        V_last = V[(n-1)*w:n*w, :]
        cross_term_old = V_old @ (S * (V_last.T @ x[(n-1)*w:n*w]))
        cross_term_new = V_last @ (S * (V_old.T @ x_old))

        result[:(n-1)*w] += cross_term_old.real
        result[(n-1)*w:n*w] += cross_term_new.real

        return result
    return LinearOperator((n*w, n*w), matvec=matvec, rmatvec=matvec, dtype=np.float64)

A11_op = make_operator(second_window[:,w:2*w], 2, w, V, S)
S11_new, V11_new = eigs(A11_op, k=2, which='LR')

print("\nStep 3 results:")
print("S11_new:", S11_new)
print("V11_new shape:", V11_new.shape)

# Step 4: Final SVD
S11_sqrt = np.sqrt(np.abs(S11_new))
S11_invsqrt = np.zeros_like(S11_new)
mask = np.abs(S11_new) > 1e-10
S11_invsqrt[mask] = 1 / np.sqrt(np.abs(S11_new[mask]))

def svd_matvec(x):
    n = 2
    x = x.reshape(-1)  # x should be of shape (2*w,)
    result = np.zeros(N, dtype=np.complex128) 
    
    # Upper block (V11 S11^(1/2) V11^T x)
    v_proj = V11_new.T @ x 
    scaled_proj = S11_sqrt * v_proj  
    result[:n*w] = V11_new @ scaled_proj  
    
    # Lower block (A21 V11 S11^(-1/2) V11^T x)
    scaled_proj_inv = S11_invsqrt * v_proj  
    temp = V11_new @ scaled_proj_inv  
    # A12^T consists of two parts:
    # 1. From V[:(n-1)w,:] S V[nw:,:]^T
    v_old_proj = V[n*w:, :] @ (np.diag(S) @ (V[:(n-1)*w, :].T @ temp[:w]))
    # 2. From new_window[:,nw:]^T 
    new_window_proj = second_window[:, n*w:].T @ temp[w:] 
    result[n*w:] = v_old_proj + new_window_proj 
    return result 

def svd_rmatvec(x): 
    n = 2
#     print(x)
    # Split input vector 
    x1, x2 = x[:n*w], x[n*w:] 
    
    # Compute A12 @ x2 efficiently 
    Vn_t_x2 = V[n*w:, :].T @ x2  # N x 1 
    scaled_Vn_t_x2 = np.diag(S) @ Vn_t_x2  # 
    A12_upper_x2 = V[:(n-1)*w, :] @ scaled_Vn_t_x2  # 2 x 1 
    A12_lower_x2 = second_window[:, n*w:] @ x2  # (w,) 
    A12_x2 = np.concatenate([A12_upper_x2, A12_lower_x2])  # (nw,) 
    
    # V11_new^T @ x1
    V11_new_t_x1 = V11_new.T @ x1  # (k,) 
    
    # V11_new^T @ (A12 @ x2)
    V11_new_t_A12_x2 = V11_new.T @ A12_x2  # (k,) 
    
    # Combine with S11 matrices - use proper diagonal matrix multiplication 
    temp = (np.diag(S11_sqrt) @ V11_new_t_x1) + (np.diag(S11_invsqrt) @ V11_new_t_A12_x2)  
    
    # Final multiplication 
    result = V11_new @ temp  # ((n+1)w,) 
    
    return result 

svd_op = LinearOperator((N, 2*w), matvec=svd_matvec, rmatvec=svd_rmatvec, dtype=np.complex128)

# Compute final SVD
U_new, S_new, Vt_new = svds(svd_op, k=keep)

print("S_new:", S_new**2)
print("U_new shape:", U_new.shape)

S = S_new**2
V = U_new

# Third window 
second_window = full_matrix[2*w:3*w, :]
A11_op = make_operator(second_window[:,w:2*w], 2, w, V, S)
S11_new, V11_new = eigs(A11_op, k=2, which='LR')

print("\nStep 3 results:")
print("S11_new:", S11_new)
print("V11_new shape:", V11_new.shape)

S11_sqrt = np.sqrt(np.abs(S11_new))
S11_invsqrt = np.zeros_like(S11_new)
mask = np.abs(S11_new) > 1e-10xv
S11_invsqrt[mask] = 1 / np.sqrt(np.abs(S11_new[mask]))

print("\nFinal results:")
print("S_new:", S_new**2)
print("U_new shape:", U_new.shape)
print("Vt_new shape:", Vt_new.shape)

svd_op = LinearOperator((N, 2*w), matvec=svd_matvec, rmatvec=svd_rmatvec, dtype=np.complex128)

# Compute final SVD
U_new, S_new, Vt_new = svds(svd_op, k=keep)


# For comparison, compute direct SVD of full matrix
U_full, S_full, Vt_full = svd(full_matrix)
print("\nTrue singular values of full matrix:", S_full)
```
